### - Limpieza de datos (Pandas: dropna, fillna, replace, etc.)

```python
df_cars.isna().sum()
# Inicialmente lo que podemos ver es que no existen valores faltantes en la base
```

```python
# Cuenta como limpieza convertir la columna Date a fecha puesto que arriba decia que era object y agregar una nueva columna con la
# con la fecha arreglada
df_cars['fecha_arr']=pd.to_datetime(df_cars['Date'],errors='coerce')
df_cars
```

```python
df_cars.info()
#Vemos que ya quedó
```

```python
#Miremos duplicados
df_cars.duplicated().sum()
# No hay duplicados
```

```python
#Finalmente identifiquemos incongruencias
df_cars[df_cars['Sale Price']<=0]
#No hay ningun carro vendido por un precio menor o igual a cero
```

### - Filtrado y selección de datos

```python
df_cars.shape
# Existen 2'500.000 autos vendidos
```

```python
sum(df_cars['Sale Price'])
#Se consiguión un monto historico de la base de 75.030.446.855 dolares
```

```python
# Ahora basandonos en las preguntas que debemos responder
# Marca que mas se vendio historicamente en la base

df_cars.groupby('Car Make')['Car Model'].count().sort_values(ascending=False)

# La marca que mas se vendio fue Honda
```

```python
# Modelo que mas se vendio historicamente en la base

df_cars.groupby('Car Model')['Car Make'].count().sort_values(ascending=False)

# La modelo que mas se vendio fue Silverado
```

```python
# Año que mas se vendieron autos
df_cars['fecha_ano']=df_cars['fecha_arr'].dt.year
df_cars
```

```python
df_cars.groupby('fecha_ano')['Car Make'].count().sort_values(ascending=False)
# Año que mas se vendieron autos 2022
```

```python
# Asesor o agente que mas comisión obtuvo
df_cars.groupby('Salesperson')['Commission Earned'].sum().sort_values(ascending=False)
# Michael Smith fue el asesor que mas comisión obtuvo 3.658.467 dolares
```

```python
# Hubo alguien que tuvo mas de una compra de vehiculos historicamente?
# Esta se puede resolver de varias formas pero lo haré con otro group by
df_cars.groupby('Customer Name')['Salesperson'].count().sort_values(ascending=False)
# En este caso si, hubieron varios clientes con mas de una compra
```

```python
conteo_vendedores= df_cars.groupby('Customer Name')['Salesperson'].count().sort_values(ascending=False)
mas1 = conteo_vendedores[conteo_vendedores > 1].count()

print(mas1)
# En total hay 283725 clientes que hicieron mas de una compra
```

```python
# Modelo de auto que mas recaudó dinero historicamente
df_cars.groupby('Car Model')['Sale Price'].sum().sort_values(ascending=False)
#Fue en este caso la silverado con 15.040.374.937 dolares
```

```python
#Cantidad de dinero neto sale price - comisión asesor
df_cars['Dinero neto']=df_cars['Sale Price']-df_cars['Commission Earned']
df_cars
```

```python
sum(df_cars['Dinero neto'])
# Ganancia neta
```

```python
a=df_cars.groupby('Car Model')['Sale Price'].sum().sort_values(ascending=False).iloc[1]
b=sum(df_cars['Sale Price'])
a/b*100
```
