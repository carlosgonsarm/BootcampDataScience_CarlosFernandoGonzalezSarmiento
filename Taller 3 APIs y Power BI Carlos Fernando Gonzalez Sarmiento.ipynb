#### Conexión al API

```python
import requests

base_url = "https://archive-api.open-meteo.com"
endpoint = "/v1/archive"

parametros = {
    "latitude": 4.7110, #Estas cordenadas son las de Bogota
    "longitude": -74.0721, #Estas cordenadas son las de Bogota
    "start_date": "2024-01-01",
    "end_date": "2025-12-31", #Ejemplo de 2024 a 2025
    "daily": [
        "temperature_2m_max",
        "temperature_2m_min",
        "temperature_2m_mean",
        "precipitation_sum",
        "wind_speed_10m_max",
        "weather_code"
    ],
    "timezone": "America/Bogota" #Que me lo traiga con la zona horaria de Bogota
}

response1 = requests.get(
    base_url + endpoint,
    params=parametros
)

if response1.status_code == 200:
    print(
        "Conexión exitosa a la API. Código de estado:",
        response1.status_code
    )
else:
    print(
        "Conexión fallida a la API. Código de estado:",
        response1.status_code
    )
```

```python
datos_json = response1.json()

datos_json["daily"]
```

#### Convertimos en API en un data frame

```python
import pandas as pd

df_clima = pd.DataFrame(datos_json["daily"])

df_clima.head()
```

### Explicación columnas
#### time: fecha.
#### temperature_2m_max: temperatura máxima.
#### temperature_2m_min: temperatura mínima.
#### temperature_2m_mean: temperatura promedio.
#### precipitation_sum: precipitación diaria en milímetros.
#### wind_speed_10m_max: velocidad máxima del viento.
#### weather_code: código que representa la condición climática.

```python
df_clima['fecha_arr']=pd.to_datetime(df_clima['time'],errors='coerce') #Convertimos columna time a fecha
df_clima
```

```python
df_clima.info() #Validamos que haya quedado convertida la nueva columna
```

```python
df_clima['Ano']=df_clima['fecha_arr'].dt.year #Extraemos y creamos una columna de año
df_clima.head()
```

```python
# Creo una variable para categorizar los dias en su nombre

dias_espanol = {
    "Monday": "Lunes",
    "Tuesday": "Martes",
    "Wednesday": "Miércoles",
    "Thursday": "Jueves",
    "Friday": "Viernes",
    "Saturday": "Sábado",
    "Sunday": "Domingo"
}

df_clima["Dia_semana"] = (
    df_clima["fecha_arr"]
    .dt.day_name()
    .map(dias_espanol)
)
df_clima.head()
```

```python
# Creo variable tomando el supuesto y apoyandome en internet de que si la precipitación es 0 entonces no hay lluvia, si es mas de 0 si la hay
import numpy as np

df_clima["Tipo_dia"] = np.where(
    df_clima["precipitation_sum"] > 0,
    "Día lluvioso",
    "Día sin lluvia"
)
df_clima.head()
```

```python
df_clima.groupby('Tipo_dia')['weather_code'].count().sort_values(ascending=False) 
# Vemos que hay días que en efector su precipitación fue 0 o no hubo lluvia
```
